```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic=sns.load_dataset('titanic')
```

```python
titanic
```

```python
plt.figure(figsize=(10,6))
```

```python
sns.boxplot(data=titanic, x='sex', y='age', hue='survived')
 #- x='sex' : Group data based on gender (male/female)
# - y='age' : Plot age distribution
# - hue='survived' : Further divide the boxes based on survival (0 = No, 1 = Yes)
plt.title('Box Plot of Age vs Sex with Survival Information')  # Title of the plot
plt.xlabel('Sex')  # Label for x-axis
plt.ylabel('Age')  # Label for y-axis

plt.show()  # Display the plot
```
