```python
import pandas as pd   # For loading and manipulating data
import seaborn as sns  # For visualization
import matplotlib.pyplot as plt  # For plotting graphs
```

```python
df=pd.read_csv('IRIS.csv')
```

```python
# Step 4: List features and their types
df.dtypes 
```

```python
# Step 5: Create histograms for each feature
df.drop('species', axis=1).hist(figsize=(10, 6), bins=20)#bins=20: Defines the number of bins for the histogram. 
#A higher number of bins gives more detailed distribution information.
plt.suptitle("Histograms for Each Feature in the Iris Dataset", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Step 6: Create boxplots for each feature
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.drop('species', axis=1))
plt.title("Boxplot for Each Feature in the Iris Dataset", fontsize=16)
plt.show()
```
